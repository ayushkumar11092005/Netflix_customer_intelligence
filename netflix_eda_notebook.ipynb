```python
import sqlite3
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Style settings
sns.set_theme(style="whitegrid")

# Database connection aur data load karna
conn = sqlite3.connect('netflix_customer_intelligence.db')
df = pd.read_sql("SELECT * FROM netflix_data", conn)
conn.close()

print("Generating all 21 Visualizations...")

# 1. Customer Age Distribution
plt.figure(figsize=(8, 4))
sns.histplot(df['age'], bins=30, kde=True, color='red')
plt.title('1. Customer Age Distribution')
plt.show()

# 2. Plan ID Count
plt.figure(figsize=(8, 4))
sns.countplot(data=df, x='plan_id', palette='Set2')
plt.title('2. Distribution of Plan IDs')
plt.show()

# 3. Subscription Status Distribution
plt.figure(figsize=(8, 4))
sns.countplot(data=df, x='subscription_status', palette='coolwarm')
plt.title('3. Subscription Status Distribution')
plt.show()

# 4. Gender Distribution
if 'gender' in df.columns:
    plt.figure(figsize=(6, 4))
    sns.countplot(data=df, x='gender', palette='pastel')
    plt.title('4. Gender Distribution')
    plt.show()

# 5. Monthly Price Distribution
if 'monthly_price' in df.columns:
    plt.figure(figsize=(8, 4))
    sns.histplot(df['monthly_price'], bins=20, kde=True, color='blue')
    plt.title('5. Monthly Price Distribution')
    plt.show()

# 6. Payment Amount Distribution
if 'amount' in df.columns:
    plt.figure(figsize=(8, 4))
    sns.histplot(df['amount'], bins=20, kde=True, color='green')
    plt.title('6. Payment Amount Distribution')
    plt.show()

# 7. Top 10 Countries by Customer Count
if 'country' in df.columns:
    plt.figure(figsize=(10, 4))
    top_countries = df['country'].value_counts().head(10)
    sns.barplot(x=top_countries.index, y=top_countries.values, palette='viridis')
    plt.title('7. Top 10 Countries by Customer Count')
    plt.xticks(rotation=45)
    plt.show()

# 8. Age Distribution across Plan IDs
if 'plan_id' in df.columns and 'age' in df.columns:
    plt.figure(figsize=(8, 4))
    sns.boxplot(data=df, x='plan_id', y='age', palette='Set3')
    plt.title('8. Age Distribution across Plan IDs')
    plt.show()

# 9. Subscription Status by Gender
if 'subscription_status' in df.columns and 'gender' in df.columns:
    plt.figure(figsize=(8, 4))
    sns.countplot(data=df, x='subscription_status', hue='gender', palette='Set1')
    plt.title('9. Subscription Status by Gender')
    plt.show()

# 10. Monthly Price by Plan ID
if 'monthly_price' in df.columns and 'plan_id' in df.columns:
    plt.figure(figsize=(8, 4))
    sns.boxplot(data=df, x='plan_id', y='monthly_price', palette='magma')
    plt.title('10. Monthly Price by Plan ID')
    plt.show()

# 11. Correlation Heatmap of Numerical Features
plt.figure(figsize=(8, 6))
num_cols = df.select_dtypes(include=[np.number]).columns
if len(num_cols) > 1:
    sns.heatmap(df[num_cols].corr(), annot=True, cmap='coolwarm', fmt='.2f')
    plt.title('11. Correlation Heatmap of Numerical Features')
    plt.show()

# 12. Customer Registrations by Year
if 'registration_date' in df.columns:
    df['reg_year'] = pd.to_datetime(df['registration_date'], errors='coerce').dt.year
    plt.figure(figsize=(8, 4))
    sns.countplot(data=df, x='reg_year', palette='rocket')
    plt.title('12. Customer Registrations by Year')
    plt.show()

# 13. Subscription Start Month Distribution
if 'subscription_start_date' in df.columns:
    df['sub_start_month'] = pd.to_datetime(df['subscription_start_date'], errors='coerce').dt.month
    plt.figure(figsize=(8, 4))
    sns.countplot(data=df, x='sub_start_month', palette='mako')
    plt.title('13. Subscription Start Month Distribution')
    plt.show()

# 14. Age Boxplot for Outliers
plt.figure(figsize=(6, 4))
sns.boxplot(y=df['age'], color='orange')
plt.title('14. Age Boxplot (Outlier Detection)')
plt.show()

# 15. Payment Amount by Subscription Status
if 'amount' in df.columns and 'subscription_status' in df.columns:
    plt.figure(figsize=(8, 4))
    sns.boxplot(data=df, x='subscription_status', y='amount', palette='Set2')
    plt.title('15. Payment Amount by Subscription Status')
    plt.show()

# 16. Top 10 Cities by Customer Count
if 'city' in df.columns:
    plt.figure(figsize=(10, 4))
    top_cities = df['city'].value_counts().head(10)
    sns.barplot(x=top_cities.index, y=top_cities.values, palette='crest')
    plt.title('16. Top 10 Cities by Customer Count')
    plt.xticks(rotation=45)
    plt.show()

# 17. Top 10 States/Regions
if 'state_or_region' in df.columns:
    plt.figure(figsize=(10, 4))
    top_states = df['state_or_region'].value_counts().head(10)
    sns.barplot(x=top_states.index, y=top_states.values, palette='flare')
    plt.title('17. Top 10 States/Regions')
    plt.xticks(rotation=45)
    plt.show()

# 18. Plan ID vs Subscription Status
if 'plan_id' in df.columns and 'subscription_status' in df.columns:
    plt.figure(figsize=(9, 4))
    sns.countplot(data=df, x='plan_id', hue='subscription_status', palette='Spectral')
    plt.title('18. Plan ID vs Subscription Status')
    plt.show()

# 19. Monthly Price Violin Plot
if 'monthly_price' in df.columns:
    plt.figure(figsize=(8, 4))
    sns.violinplot(y=df['monthly_price'], color='purple')
    plt.title('19. Monthly Price Violin Plot')
    plt.show()

# 20. Cumulative Age Distribution (ECDF)
plt.figure(figsize=(8, 4))
sns.ecdfplot(df['age'], color='brown')
plt.title('20. Cumulative Age Distribution (ECDF)')
plt.show()

# 21. Payment Amount Violin Plot across Plan IDs
if 'amount' in df.columns and 'plan_id' in df.columns:
    plt.figure(figsize=(8, 4))
    sns.violinplot(data=df, x='plan_id', y='amount', palette='Pastel1')
    plt.title('21. Payment Amount Violin Plot across Plan IDs')
    plt.show()

print("All 21 Visualizations Generated Successfully!")
```
